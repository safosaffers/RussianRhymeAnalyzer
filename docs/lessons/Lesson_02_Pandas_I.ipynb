## Объект `Series`

```python
a = [10, 20, 30] # это список (list)
print(a)
```

```python
print(pd.Series(a)) # а это ряд (Series)
```

В выводе два столбца: индексы и данные. `Series` (ряд) похож на массив с индексами, причем можно использовать различные метки индексов (а не только целые числа, начиная с 0). То есть он похож на словарь. Ряд в отличие от базовых типов данных — словарей и списков — содержит данные только одного типа, который можно задать.

```python
s = pd.Series(a, index=['a', 'b', 'c']) # создадим новый ряд, явно укажем метки для индексирования
print(s)
```

```python
# извлекаем один и тот же элемент ряда
print(s[0])  # s.iloc[0] - правильнее
print(s['a'])
```

```python
print(s[['a', 'c']])
```

```python
print(s[:'c']) # делаем срез
```

```python
print(s[:2])
```

Посмотрим на индексы объекта `s`.

```python
print(s.index)
```

Значения ряда можно получить с помощью свойства `values`.

```python
print(s.values)
```

Получим информацию об объекте.

```python
print(len(s))  # количество элементов (по нулевой оси)
print(s.shape) # количество строк и столбцов
```

Над объектами Series можно совершать различные арифметические операции (`+`, `-`, `/`, `*`, ...). Они будут выполняться поэлементно в случае совпадения индексов.

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

s1 + s2
```

```python
# если ряды содержат разное количество элементов,
# то операция выполняется только для существующих элементов
s1 = pd.Series([1, 2, 3, 4, 5, 6, 7, 8, 9])
s2 = pd.Series([10, 20, 30, 40, 50])

s1 + s2
```

```python
# если ряды имеют разные метки индексов,
# то операция выполняется только для элементов с соответствующими индексами
s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s2 = pd.Series([10, 20, 30, 40, 50])

s1 + s2
```

## Описательная статистика

Для объектов Series можно вычислить различные статистики, используя методы

* `count()` — количество непропущенных значений   
* `nunique()` — количество уникальных значений
* `unique()` — уникальные значения
* `value_counts()` — уникальные значения и их количества
* `mode()` — мода
* `mean()` — среднее арифметическое
* `median()` — медиана
* `sum()`  
* `min()` и `max()`

```python
s = pd.Series([10, 20, 30, 30, 30, 40, 50])

print('count =', s.count(), end='\n\n')
print('nunique =', s.nunique(), end='\n\n')
print('unique =', s.unique(), end='\n\n')
print('value_counts:\n', s.value_counts(), end='\n\n', sep='')
print('mode =', s.mode()[0], end='\n\n')
print('mean =', s.mean(), end='\n\n')
print('median =', s.median(), end='\n\n')
```

## Объект `DataFrame`

Объект `DataFrame` (датафрейм) представляет собой один или несколько объектов `Series`, упорядоченных по меткам идекса. Каждый ряд будет столбцом в датафрейме со своим именем.

Датафрейм подобен таблице реляционной базы данных, поскольку содержит столбцы разных типов, но данные в каждом столбце имеют один и тот же тип.

```python
df = pd.DataFrame({'letters':ser1, 'floats':ser2, 'integers':s})
print(df)
```

```python
ser2 = ser2.reset_index(drop=True)
df = pd.DataFrame({'letters':ser1, 'floats':ser2, 'integers':s})
print(df)
```

### Извлечение данных из датафрейма

Получим данные из определенного столбца (он же признак или атрибут).

```python
print(df['letters']) # это ряд
```

```python
print(df[['letters', 'integers']]) # несколько столбцов
```

Извлекать строки из датафрейма можно с помощью методов `loc` и `iloc`.

Метод `loc` ищет по метке индекса, `iloc` — осуществляет поиск по позиции (начиная с 0).

```python
print(df.iloc[:, 0]) # получим все строки столбца letters
```

```python
print(df.loc[:, 'letters']) # то же
```

```python
print(df.iloc[:3, :2]) # получим первые три строки первых двух рядов
```

```python
print(df.loc[:2, ['letters', 'floats']]) # то же
```

```python
print(df[:3][['letters', 'floats']]) # то же
```

Строки датафрейма можно отбирать по условию.

```python
print(df['integers'] < 30)
```

```python
print(df[df['integers'] < 30])
```

 ## Конкатенация данных

Конкатенацию данных из разных объектов выполняют функцией `pd.concat()`. Конкатенация объектов `Series` приводит к образованию нового объекта `Series` с объединенными значениями.

```python
s1 = pd.Series([0, 1, 2])
s2 = pd.Series([3, 4, 5])

pd.concat([s1, s2])
```

Получили новых объект `Series` с дублирующими метками индексов. В этом случае помогает сброс индекса:

```python
pd.concat([s1, s2]).reset_index(drop=True)
```

Можно также игнорировать метки индекса в исходных объектах:

```python
pd.concat([s1, s2], ignore_index=True) # тот же результат
```

### Конкатенация датафреймов

Объекты `DataFrame` конкатенируются аналогичным образом.

```python
df1 = pd.DataFrame({'a': s1, 'b': s2})
df1
```

```python
df2 = pd.DataFrame({'a': s1 + 10, 'b': s2 + 10})
df2
```

Выполним конкатенацию датафреймов:

```python
pd.concat([df1, df2])
```

По умолчанию добавляются новые строки. Значения в соответствующих столбцах идут по порядку, указанному в функции `pd.concat()`. Здесь мы тоже получили дублирующие метки индекса строк.  

Рассмотрим вариант с несовпадающими названиями колонок объединяемых датафреймов.

```python
df3 = pd.DataFrame({'b': s1 * -1, 'c': s2 * -1})
pd.concat([df1, df3]) # совпадает только колонка b
```

В итоговом датафрейме отсутствующие значения столбцов заполнены `NaN` значениями.

Каждой группе данных в итоговом датафрейме можно дать свое название, используя параметр `keys`. Он создаст иерархический индекс строк:

```python
res = pd.concat([df1, df3], keys=['df1', 'df3'])
res
```

```python
res.loc['df1'] # извлекаем данные из итоговой таблицы, соответствующие первому датафрейму
```

Функция `pd.concat()` позволяет задать ось, по которой произойдет добавление новых данных. По умолчанию добавляются строки (`axis=0`). Конкатенируем два объекта по оси столбцов, указав параметр `axis=1`:

```python
pd.concat([df1, df3], axis=1) # получим дублирующиеся наименования столбцов
```

Если метки индекса строк будут разными, то получим `NaN` значения:

```python
df4 = pd.DataFrame({'a': [0, 1, 2], 'b': [3, 4, 5]},
                   index=[2, 3, 4])
df4 # с df1 совпадает только метка 2
```

```python
pd.concat([df1, df4], axis=1)
```

Индексы строк у исходных датафреймов совпадают только меткой 2. Остальные метки не совпадают, поэтому в итоговом датафрейме стоят `NaN` значения.

### Тип соединения данных

По умолчанию конкатенация выполняет операцию **внешнего соединения** (outer join) индексных меток. Тип соединения можно изменить на **внутреннее соединение** (inner join), указав в качестве параметра `join='inner'`. В этом случае в итоговой таблице останутся только те строки, у которых совпадает метка:

```python
# выполним внутреннее соединение индексных меток по оси строк
pd.concat([df1, df4], axis=1, join='inner') # останется строка с меткой 2
```

Выполним конкатенацию вдоль оси строк (`axis=0`), или другими словами - внешнее и внутреннее соединение индексных меток по оси столбцов.

```python
# выполним внешнее соединение индексных меток по оси столбцов
pd.concat([df1, df3], axis=0, join='outer') # получим все три столбца
```

```python
# выполним внутреннее соединение индексных меток по оси столбцов
pd.concat([df1, df3], axis=0, join='inner') # останется столбец с меткой b
```

## Слияние данных

Функция `pd.merge()` и метод `.merge()` выполняют соединение датафреймов аналогично операции соединения для баз данных. В результате получим новый объект, который содержит объединенные данные двух объектов по совпадающим значениям в индексах строк или столбцов.

```python
# создадим датафрейм с авторами
authors = {
    'id_author': [1, 2, 3, 4, 5],
    'name': ['Ротфусс', 'Герберт', 'Хобб', 'Сандерсон', 'Пратчетт']
}
authors = pd.DataFrame(authors)
authors
```

```python
# создадим датафрейм с названиями книг и ID авторов
books = {
    'title': ['Имя ветра', 'Стража! Стража!', 'Шляпа, полная неба', 'Путь королей', 'Темная башня'],
    'id_author': [1, 5, 5, 4, 10]
}
books = pd.DataFrame(books)
books
```

Соединим датафреймы `books` и `authors` по общей колонке `id_author`. Метод `.merge()` применяет внутреннее соединение (`inner join`) по умолчанию.

```python
books.merge(authors)
```

Выше метод `.merge()` сам определил общую колонку по названию. Но мы всегда можем явно указать имя столбика, по которому нужно выполнить соединение в параметре `on`. Столбец, указанный в этом параметре, должен присутствовать в каждом датафрейме.

```python
books.merge(authors, on='id_author')
```

Мы также можем изменить тип соединения в параметре `how`. Допустимо использовать следующие значения параметра:

• `inner`: выполняет пересечение по соответствующим значениям общего столбика  
• `outer`: выполняет объединение строк из обоих датафреймов   
• `left`: присоединяет соответствующие строки к левому датафрейму из правого  
• `right`: присоединяет соответствующие строки к правому датафрейму из левого

```python
books.merge(authors, how='outer') # внешнее соединение books и authors
```

```python
authors.merge(books, how='outer') # внешнее соединение authors и books
```

```python
authors.merge(books, how='left') # левое соединение authors и books
```

```python
authors.merge(books, how='right') # правое соединение authors и books
```
