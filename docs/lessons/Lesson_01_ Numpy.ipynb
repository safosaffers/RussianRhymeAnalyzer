# Библиотека NumPy

Содержит константы, функции и объекты для работы с многомерными массивами данных. Используется в анализе данных, машинном обучении, научных вычислениях.

```python
import numpy as np
```

## Создание массивов

Основным объектом NumPy является многомерный массив (numpy.ndarray). Это n-мерный массив из элементов (обычно чисел) одного типа - вектор, матрица, тензор и т.д. Используются для представления табличных данных, изображений, текста, звука.

### Одномерные массивы

Функция `array()` трансформирует последовательность в многомерный массив.

```python
a = np.array([1, 2, 3])
print(type(a))
print(a)
```

```python
a = ['AAA', 1, 2, 3, 4, 5]
print(np.array(a))
```

Функция `arange()`, аналогичная встроенной в Python `range()`, возвращает массив.

```python
a = np.arange(2, 20, 3) # создание массива в диапазоне [2, 20) с шагом 3
print(a)
```

Функция `linspace(start, stop, num)` создает массив размером `num` элементов из чисел в диапазоне от `start` до `stop` с равным промежутком.

```python
print(np.linspace(100, 1000, 10))
```

Функция `logspace(start, stop, num, base=10)` создает массив размером `num` элементов из чисел в диапазоне от `base ** start` до `base ** stop` с равным промежутком.

```python
print(np.logspace(3., 10., 10))
```

Функция `zeros()` создает массив из нулей, функция `ones()` — массив из единиц. Функция `eye()` создаёт единичную матрицу. Эти функции принимают кортеж с размерами, и аргумент dtype.

```python
zeros = np.zeros((2, 3), dtype=np.int32)
ones = np.ones((5, 10), dtype=np.float64)
identity = np.eye(3)

print(zeros, end='\n\n')
print(ones, end='\n\n')
print(identity)
```

## Индексация и срезы

Одномерные массивы похожи на простые списки Python. Обратиться к элементу можно по его индексу.

```python
a = np.arange(10)
a[2]
```

Обращение к элементам и извлечение срезов в многомерных массивах производится не по одному, а по нескольким индексам (каждый индекс соответствует своей оси).

```python
print(a2)
print('Отдельные элементы: ', a2[0, 0], a2[1, 1], a2[-1, -1])
```

Можно создать маску индексов и по маске вытащить нужные элементы.

```python
evens = np.array([x for x in range(10) if x % 2 == 0])
print(evens) # маска индексов
letters = np.array(list('abcdefghij'))
print(letters)
print(letters[evens])
```

```python
indexes = np.array([True] * 5 + [False] * 5)
print(indexes) # маска индексов
print(letters[indexes])
```

## Соединение массивов

```python
a = np.zeros(9, dtype=int).reshape(3, 3)
b = np.ones(9, dtype=int).reshape(3, 3)

ab1 = np.hstack((a, b))
ab2 = np.vstack((a, b))

print('a:\n', a)
print('b:\n', b)
print('Horizontal Stack:\n', ab1)
print('Vertical Stack:\n', ab2)
```

```python
a = np.ones(16).reshape(2, -1)

a_1, a_2 = np.vsplit(a, 2) # разделяем вертикально на 2 массива
# a_1, a_2 = np.hsplit(a, 2)

print(a, end='\n\n')
print(a_1, end='\n\n')
print(a_2, end='\n\n')
```

```python
print(np.r_[0:10, 4:8, 100]) # горизонтальное соединение - в строку
```

```python
print(np.c_[1:3, 7:9, 0:2]) # вертикальное соединение - в столбцы
```

```python
a = np.arange(1, 10)
print(np.split(a, [5])) # разделение массива по элементу
print(np.split(a, [4, 6]))
```

## Сортировка массивов

```python
a = np.array([[1,4],[3,1]])
print(a, end='\n\n')
print(np.sort(a, axis=None), end='\n\n')  # сортировка одномерного массива
print(np.sort(a, axis=0), end='\n\n')     # сортировка по столбцам
print(np.sort(a, axis=1), end='\n\n')     # сортировка по строкам
```

## Математические операции

Математические операции над массивами выполняются поэлементно. Создается новый массив, который заполняется результатами действия оператора.

```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** b)
```

Массивы должны быть одинаковых размеров.

```python
a = np.arange(9)
b = np.arange(1)
print(a + b)
```

Математические операции между массивом и числом.

```python
print(a + 100)
print(a ** 2)
print(a > 5)
```

NumPy также предоставляет множество математических операций для обработки массивов.

```python
print(np.sin(a))
print(np.exp(a))
print(np.log1p(a))
print(np.sqrt(a))
```

## Агрегирование массивов

Многие унарные операции, такие как `np.sum()`, представлены также и в виде методов класса `ndarray`.

```python
print(np.sum(a))
print(a.sum())
```

По умолчанию эти операции применяются к массиву целиком, независимо от его формы. Однако, указав параметр `axis`, можно применить операцию для указанной оси массива.

```python
a = a.reshape(3, -1)
print(a)
print('Min по столбцам:\n', a.min(axis=0))
print('Min по строкам:\n', a.min(axis=1))
print('Min по матрице:\n', a.min())
```

```python
print(a.max(axis=1))  # максимальные значения по строкам
print(a.sum(axis=1))  # сумма по строкам
print(a.prod(axis=1)) # произведение по строкам
print(a.mean(axis=1)) # среднее значение по строкам
```

Функции `argmax()` и `argmin()` возвращают индексы максимальных или минимальных элементов по осям.

```python
print(a.argmax(axis=0))
print(a.argmax(axis=1))
```

## Линейная алгебра
Модуль библиотеки `numpy.linalg` позволяет выполнять многие операции из линейной алгебры.

```python
a = np.arange(1, 5).reshape(2, 2)
print(a)
np.linalg.det(a).round() # определитель матрицы
```

### Умножение матриц
Умножать можно только матрицы, отвечающие следующему требованию: количество столбцов первой матрицы должно быть равно числу строк второй матрицы.

```python
a = np.arange(1, 9).reshape(2, 4)
b = np.arange(8, 0, -1).reshape(4, 2)
print(a)
print(b)
```

```python
c = np.dot(a, b) # скалярное произведение матриц
print(c)
```

```python
a @ b
```

### Обратная матрица

```python
c_inv = np.linalg.inv(c)
print(c_inv)
```

```python
print(np.round(np.dot(c, c_inv)))
```

## Генерация случайных чисел в numpy
Для создания массивов со случайными элементами служит модуль `numpy.random`.

```python
np.random.seed(1000) # инициализация генератора псевдослучайных чисел
```

Функция `sample()` возвращает числа из промежутка `[0, 1)` в виде константы, вектора или матрицы.

```python
np.random.seed(1000)
print(np.random.sample(10))
```

```python
print(np.random.sample((2, 3)))
```

Функция `randint(low, high, size)` создает массив из целых чисел. Параметры `low, high, size`: от какого, до какого числа, размер массива.

```python
print(np.random.randint(0, 5, 10))
```

Функция `uniform(low, high, size)` создает массив из вещественных чисел, используя равномерное распределение. Параметры `low, high, size`: от какого, до какого числа, размер массива.

```python
print(np.random.uniform(0, 5, (3, 3)))
```

Функция `rand(d1, d2, d3, ...)` получает требуемый размер массива и заполняет его числами, которые случайно распределены в интервале от 0 до 1.

```python
print(np.random.rand(3, 3, 3))
```

Функция `randn(d1, d2, d3, ...)` создает массив указанной формы и заполняет его случайными числами с плавающей точкой, которые выбраны из одномерного нормального распределения со средним значением равным 0 и дисперсией равной 1.

```python
print(np.random.randn(3, 5))
```

Перемешать массив можно с помощью функций `shuffle()` или `permutation()`.

```python
a = np.arange(10)
np.random.shuffle(a)
print(a)
```

```python
print(np.random.permutation(a)) # создает копию
```

```python
print(a)
```

Сделать случайную выборку из массива можно с помощью функции `choice()`.

```python
print(np.random.choice(a, 10, replace=True))
```

## Сравнение времени выполнения кода

```python
def sum_for(array1, array2):
  sum = 0
  for el1, el2 in zip(array1, array2):
    sum += el1 + el2
  return sum

n = 1000000
lst1 = [i for i in range(n)]
lst2 = [i for i in range(n)][::-1]
%timeit sum_for(lst1, lst2)
```

```python
import numpy as np

def sum_numpy(array1, array2):
  return np.sum(array1 + array2)

arr1 = np.array(lst1)
arr2 = np.array(lst2)
%timeit sum_numpy(arr1, arr2)
```
