# Библиотека Pandas


Pandas — самая популярная библиотека Python для обработки и анализа данных. Предлагает высокопроизводительные структуры данных и инструменты, которые решают следующие задачи:

* **загрузка данных** из внешних источников (электронный таблиц, баз данных, веб-страниц и прочих разных)
* **сохранение данных** в требуемом формате
* объединение данных, полученных из разных источников
* очистка данных от пропусков и ошибок
* **группировка и агрегирование данных**
* преобразование данных
* выполнение запросов для исследования данных
* создание новых атрибутов/признаков данных

## Группировка данных

`DataFrame.groupby(by = grouping_columns)[columns_to_show].function()`



Найдем медиану рейтинга фильмов каждого режиссера.

```python
df.groupby('Director')['Rating'].median()
```

Найдем средний рейтинг фильмов каждого жанра и отсортируем по убыванию.

```python
df.groupby('Genre')['Rating'].mean().sort_values(ascending=False).head(3)
```

Выведем данные одной группы - get_group()

```python
df.groupby('Genre').get_group('Action,Adventure')[['Genre', 'Title', 'Rating']]
```

Если надо для каждой группы посчитать не одну функцию.

```python
df.groupby('Genre')['Rating'].agg(['min', 'mean', 'max'])
```

```python
df.groupby('Genre')['Rating'].agg(
      rating_min='min',
      rating_mean='mean',
      rating_max='max'
)
```

Если считаем не по одному столбцу.

```python
df.groupby('Director').agg(
    num_films=('Title', 'count'),
    votes_mean=('Votes', 'mean'),
    revenue_sum=('Revenue (Millions)', 'sum')
).sort_values('num_films', ascending=False)
```

## Сводные таблицы

```python
pd.pivot_table(data=df,
               index='Director',
               values='Votes',
               aggfunc='mean'
)
```

```python
pd.pivot_table(data=df,
               index='Director',
               values=['Title', 'Votes', 'Revenue (Millions)'],
               aggfunc={'Title': 'count',
                        'Votes': 'mean',
                        'Revenue (Millions)': 'sum'}
               ).sort_values('Title', ascending=False)
```

```python
pd.pivot_table(data=df,
               index='Director',
               values=['Title', 'Votes', 'Revenue (Millions)'],
               aggfunc={'Title': 'count',
                        'Votes': 'mean',
                        'Revenue (Millions)': 'sum'}
               ).sort_values('Title', ascending=False).rename(
    columns={'Title': 'num_films',
             'Votes': 'votes_mean',
             'Revenue (Millions)': 'revenue_sum'}
)
```

## Сохранение DataFrame в файл

```python
df_gr = df.groupby('Director').agg(
    num_films=('Title', 'count'),
    votes_mean=('Votes', 'mean'),
    revenue_sum=('Revenue (Millions)', 'sum')
).sort_values('num_films', ascending=False)

df_gr.to_csv('/content/IMDB-Movie-Data-gr.csv')
```
